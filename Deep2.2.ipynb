5. Design and implement a CNN model (with 4+ layers of
convolutions) to classify multi category image datasets. Use
the concept of regularization and dropout while designing the
CNN model. Use the Fashion MNIST datasets. Record the
Training accuracy and Test accuracy corresponding to the
following architectures: a. Base Model b. Model with L1
Regularization c. Model with L2 Regularization d. Model with
Dropout e. Model with both L2 (or L1) and Dropout

```python
#5 models ka sangam
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.regularizers import l1, l2

# 1. Load and Preprocess Fashion MNIST
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
X_train = X_train.reshape(-1, 28, 28, 1) / 255.0
X_test = X_test.reshape(-1, 28, 28, 1) / 255.0

# 2. Define Clean Model Builder Function
def create_model(reg=None, use_dropout=False):
    model = Sequential([
        # Conv Layer 1 & 2
        Conv2D(32, (3, 3), activation="relu", padding="same", input_shape=(28, 28, 1)),
        Conv2D(32, (3, 3), activation="relu", padding="same"),
        MaxPooling2D((2, 2)),
        
        # Conv Layer 3 & 4 (4+ convolution layers total)
        Conv2D(64, (3, 3), activation="relu", padding="same"),
        Conv2D(64, (3, 3), activation="relu", padding="same"),
        MaxPooling2D((2, 2)),
        
        # Dense Classifier Layers
        Flatten(),
        Dense(128, activation="relu", kernel_regularizer=reg)
    ])
    
    if use_dropout:
        model.add(Dropout(0.5))
        
    model.add(Dense(10, activation="softmax"))
    
    model.compile(
        optimizer="adam", 
        loss="sparse_categorical_crossentropy", 
        metrics=["accuracy"]
    )
    return model

# 3. Dictionary of Required Architectures
models = {
    "Base": create_model(),
    "L1": create_model(reg=l1(0.001)),
    "L2": create_model(reg=l2(0.001)),
    "Dropout": create_model(use_dropout=True),
    "L2 + Dropout": create_model(reg=l2(0.001), use_dropout=True)
}

# 4. Train and Evaluate Each Model
for name, model in models.items():
    print(f"\nTraining Model: {name}")
    model.fit(X_train, y_train, epochs=5, batch_size=128, verbose=0)
    
    train_acc = model.evaluate(X_train, y_train, verbose=0)[1]
    test_acc = model.evaluate(X_test, y_test, verbose=0)[1]


    print("Training Accuracy:", train_acc)
    print("Test Accuracy:", test_acc)
```

6. Design and implement a CNN model to classify CIFAR10 image
dataset. Use the concept of Data Augmentation while designing
the CNN model. Record the accuracy corresponding to the
number of epochs.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. Load & Scale CIFAR-10 (and flatten labels)
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()
#normalize
X_train = X_train / 255.0
X_test = X_test / 255.0
#flatten
y_train = y_train.flatten()
y_test = y_test.flatten()

# 2. Data Augmentation
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)
datagen.fit(X_train)

# 3. Build CNN Model
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(32,32,3)),
    MaxPooling2D(),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

# 4. Compile
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# 5. Train with Augmentation
model.fit(datagen.flow(X_train, y_train, batch_size=128), epochs=5, validation_data=(X_test, y_test))

# 6. Evaluate
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test Accuracy:", test_acc)
```

```python
# Q7 - Nature 12K / iNaturalist CNN
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Activation 

# Variables
img_size = (224, 224, 3)
num_classes = 10
filters = [16, 32, 64, 128, 256]
kernel_size = (3, 3)
activation_func = "relu"
dense_neurons = 256

# CNN Model
model = Sequential()

# Block 1
model.add(Conv2D(filters[0], kernel_size, input_shape=img_size))
model.add(Activation(activation_func))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Block 2
model.add(Conv2D(filters[1], kernel_size))
model.add(Activation(activation_func))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Block 3
model.add(Conv2D(filters[2], kernel_size))
model.add(Activation(activation_func))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Block 4
model.add(Conv2D(filters[3], kernel_size))
model.add(Activation(activation_func))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Block 5
model.add(Conv2D(filters[4], kernel_size))
model.add(Activation(activation_func))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Dense Layers
model.add(Flatten())
model.add(Dense(dense_neurons, activation=activation_func))
model.add(Dense(num_classes, activation="softmax"))

# Compile & Summary
model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
model.summary()
```
