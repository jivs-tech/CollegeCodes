````python
from textblob import TextBlob

# Load the tweets
tweets = [
    "I love this phone.",
    "This movie is boring.",
    "The weather is okay today.",
    "Amazing service and friendly staff!",
    "Worst app ever."
]

# Analyze each tweet
for tweet in tweets:
    blob = TextBlob(tweet)
    polarity = blob.sentiment.polarity

    # Classify sentiment
    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    # Print results
    print("Tweet      :", tweet)
    print("Polarity   :", polarity)
    print("Sentiment  :", sentiment)
    print("-" * 50)
````

````python
import nltk

nltk.download('vader_lexicon')
````

````python
from nltk.sentiment import SentimentIntensityAnalyzer

# Initialize VADER
sia = SentimentIntensityAnalyzer()

# Load tweets
tweets = [
    "Excellent service.",
    "The exam was horrible.",
    "The weather is okay.",
    "I absolutely love ChatGPT!",
    "Very disappointing experience."
]

# Analyze each tweet
for tweet in tweets:
    scores = sia.polarity_scores(tweet)

    print("Tweet      :", tweet)
    print("Negative   :", scores['neg'])
    print("Neutral    :", scores['neu'])
    print("Positive   :", scores['pos'])
    print("Compound   :", scores['compound'])

    # Classify sentiment
    if scores['compound'] >= 0.05:
        sentiment = "Positive"
    elif scores['compound'] <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Sentiment  :", sentiment)
    print("-" * 50)
````

Tasks
____________________________________________
1.Create a DataFrame.
2.Apply TF-IDF Vectorization.
3.Train a Multinomial Naive Bayes classifier.
4.Predict the sentiment for:
"Excellent battery"
5.Print the predicted sentiment.
_____________________________________________


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

# Step 1: Create a DataFrame
data = {
    'text': [
        'Good phone',
        'Excellent camera',
        'Bad battery',
        'Worst service',
        'Amazing performance',
        'Poor quality'
    ],
    'label': [
        'Positive',
        'Positive',
        'Negative',
        'Negative',
        'Positive',
        'Negative'
    ]
}

df = pd.DataFrame(data)

# Step 2: Apply TF-IDF Vectorization
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["text"])   # Correct column name
y = df["label"]                            # Correct column name

# Step 3: Train a Multinomial Naive Bayes classifier
model = MultinomialNB()
model.fit(X, y)

# Step 4: Predict the sentiment
new_text = ["Excellent battery"]
new_text_tfidf = vectorizer.transform(new_text)

prediction = model.predict(new_text_tfidf)

# Step 5: Print the predicted sentiment
print("Text:", new_text[0])
print("Predicted Sentiment:", prediction[0])
````

```
Tasks
___________________________________________________
1.Split the dataset into training and testing sets.
2.Apply TF-IDF.
3.Train a MultinomialNB model.
4.Predict the test data.
5.Print:
Actual labels
Predicted labels
Accuracy
___________________________________________________

````python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Step 1: Create DataFrame
data = {
    'text':[
        'Good product',
        'Excellent quality',
        'Poor battery',
        'Worst experience',
        'Amazing phone',
        'Bad service'
    ],
    'label':[
        'Positive',
        'Positive',
        'Negative',
        'Negative',
        'Positive',
        'Negative'
    ]
}
df = pd.DataFrame(data)

# Step 2: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    df['text'],
    df['label'],
    test_size=0.33,
    random_state=42
)

# Step 3: Apply TF-IDF
vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# Step 4: Train Multinomial Naive Bayes
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

# Step 5: Predict the test data
y_pred = model.predict(X_test_tfidf)

# Step 6: Print results
print("Actual Labels:")
print(list(y_test))

print("\nPredicted Labels:")
print(list(y_pred))

print("\nAccuracy:")
print(accuracy_score(y_test, y_pred))
````

````python
from textblob import TextBlob

# Sample reviews
reviews = [
    "The camera quality is excellent.",
    "Battery backup is poor.",
    "Display is fantastic.",
    "The phone hangs frequently.",
    "Worth every penny.",
    "Waste of money."
]
# Counters
positive = 0
negative = 0
neutral = 0

# Analyze each review
for review in reviews:
    blob = TextBlob(review)
    polarity = blob.sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
        positive += 1
    elif polarity < 0:
        sentiment = "Negative"
        negative += 1
    else:
        sentiment = "Neutral"
        neutral += 1

    print("Review     :", review)
    print("Polarity   :", polarity)
    print("Sentiment  :", sentiment)
    print("-" * 50)

# Print total counts
print("\nTotal Positive Reviews:", positive)
print("Total Negative Reviews:", negative)
print("Total Neutral Reviews :", neutral)
````

1.Perform sentiment analysis using TextBlob.
2.Perform sentiment analysis using VADER.
3.Compare the outputs.
______________________________________________

````python
from textblob import TextBlob
from nltk.sentiment import SentimentIntensityAnalyzer
import nltk

# Download VADER lexicon (Run only once)
nltk.download('vader_lexicon')

# Initialize VADER
sia = SentimentIntensityAnalyzer()

# Sample tweets
tweets = [
    "Amazing product",
    "Worst customer support",
    "Very good quality",
    "Terrible experience",
    "The delivery was okay."
]

print("{:<30} {:<12} {:<12}".format("Tweet", "TextBlob", "VADER"))
print("-" * 60)

for tweet in tweets:

    # ---------- TextBlob ----------
    tb_polarity = TextBlob(tweet).sentiment.polarity

    if tb_polarity > 0:
        tb_sentiment = "Positive"
    elif tb_polarity < 0:
        tb_sentiment = "Negative"
    else:
        tb_sentiment = "Neutral"

    # ---------- VADER ----------
    vader_score = sia.polarity_scores(tweet)

    if vader_score['compound'] >= 0.05:
        vader_sentiment = "Positive"
    elif vader_score['compound'] <= -0.05:
        vader_sentiment = "Negative"
    else:
        vader_sentiment = "Neutral"

    # ---------- Output ----------
    print("{:<30} {:<12} {:<12}".format(
        tweet,
        tb_sentiment,
        vader_sentiment
    ))
````
