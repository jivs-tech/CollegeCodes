```python
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer

# Create stemmer objects
p = PorterStemmer()
s = SnowballStemmer('english')
l = LancasterStemmer()

# Words to stem
words = ["studies", "running", "happiness", "organisation"]

# Compare all three stemmers
for w in words:
    print(f"{w:>12} | porter={p.stem(w):<9} snowball={s.stem(w):<9} lancaster={l.stem(w)}")
```

```python
from nltk.stem import WordNetLemmatizer
lem = WordNetLemmatizer()
words = ["studies", "running", "geese", "better"]
print("default (noun):", [lem.lemmatize(w) for w in words])
print("as verbs :", [lem.lemmatize(w, pos='v') for w in words])
print("'better' adj :", lem.lemmatize("better", pos='a'))
```

```python
import re
import emoji
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Chat-word dictionary
chat_words = {
    "IDK": "I do not know",
    "BTW": "by the way",
    "LOL": "laughing out loud",
    "ASAP": "as soon as possible",
    "BRB": "be right back"
}

# Function to expand chat words
def expand_chat(text):
    return " ".join(chat_words.get(w.upper(), w) for w in text.split())

# Lemmatizer object
lem = WordNetLemmatizer()

# Complete preprocessing function
def preprocess(text, for_sentiment=False):

    # 2. Lowercase
    text = text.lower()

    # 3. Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)

    # 4. Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # 5. Expand chat words
    text = expand_chat(text)

    # 6/7. Remove emojis
    text = emoji.replace_emoji(text, replace=' ')

    # 8. Remove punctuation
    text = re.sub(r'[^\w\s]', ' ', text)

    # 9. Remove numbers
    text = re.sub(r'\d+', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # 10. Tokenize
    tokens = word_tokenize(text)

    # 11. Stopword removal
    stop_en = set(stopwords.words('english'))

    # Keep negation words for sentiment analysis
    if for_sentiment:
        stop_en -= {"not", "no", "nor"}

    tokens = [w for w in tokens if w not in stop_en]

    # 13. Lemmatization
    tokens = [lem.lemmatize(w) for w in tokens]

    return tokens

# Sample text
sample = "Hey BTW visit https://shop.com 😍 I do NOT like the 3 ads!!! <b>ok</b>"

print("RAW :", sample)
print("CLEAN (general) :", preprocess(sample))
print("CLEAN (sentiment):", preprocess(sample, for_sentiment=True))
```
