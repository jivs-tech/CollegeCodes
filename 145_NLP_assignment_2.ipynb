```python
import nltk
import spacy
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer

# Download tokenizer (run once)
nltk.download('punkt')

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Input text
text = "Doctors were analyzing symptoms more carefully than usual."

# Tokenize
tokens = word_tokenize(text.lower())

# ----------------------------
# Porter Stemmer
# ----------------------------
porter = PorterStemmer()
porter_output = [porter.stem(word) for word in tokens]

# ----------------------------
# Snowball Stemmer
# ----------------------------
snowball = SnowballStemmer("english")
snowball_output = [snowball.stem(word) for word in tokens]

# ----------------------------
# spaCy Lemmatizer
# ----------------------------
doc = nlp(text)
lemma_output = [token.lemma_ for token in doc]

# ----------------------------
# Display Results
# ----------------------------
print("Original Tokens:")
print(tokens)

print("\nPorter Stemmer:")
print(porter_output)

print("\nSnowball Stemmer:")
print(snowball_output)

print("\nspaCy Lemmatizer:")
print(lemma_output)
```

```python
import re
import string
from bs4 import BeautifulSoup
import nltk
import spacy
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Download NLTK resources (Run once)
#nltk.download('punkt')
#nltk.download('stopwords')

# Load spaCy model (Run: python -m spacy download en_core_web_sm)
nlp = spacy.load("en_core_web_sm")

# Input text
text = "This drug caused side effects !!! Visit http://med.com Doctors wared about overdosed </div>."

print("Original Text:")
print(text)

# -----------------------------
# 1. Lowercase
# -----------------------------
text = text.lower()
print("\nLowercase:")
print(text)

# -----------------------------
# 2. Remove HTML Tags
# -----------------------------
text = BeautifulSoup(text, "html.parser").get_text()
print("\nAfter Removing HTML:")
print(text)

# -----------------------------
# 3. Remove URLs
# -----------------------------
text = re.sub(r'https?://\S+|www\.\S+', '', text)
print("\nAfter Removing URL:")
print(text)

# -----------------------------
# 4. Remove Punctuation
# -----------------------------
text = text.translate(str.maketrans('', '', string.punctuation))
print("\nAfter Removing Punctuation:")
print(text)

# -----------------------------
# 5. Tokenize
# -----------------------------
tokens = word_tokenize(text)
print("\nTokens:")
print(tokens)

# -----------------------------
# 6. Remove Stopwords
# -----------------------------
stop_words = set(stopwords.words('english'))
tokens = [word for word in tokens if word not in stop_words]
print("\nAfter Stopword Removal:")
print(tokens)

# -----------------------------
# 7. Stemming
# -----------------------------
porter = PorterStemmer()
stems = [porter.stem(word) for word in tokens]
print("\nPorter Stemmer:")
print(stems)

# -----------------------------
# 8. Lemmatization
# -----------------------------
doc = nlp(" ".join(tokens))
lemmas = [token.lemma_ for token in doc]
print("\nspaCy Lemmatization:")
print(lemmas)

# =====================================================
# Bag of Words
# =====================================================
print("\n========== BAG OF WORDS ==========")

cv = CountVectorizer()
bow = cv.fit_transform([" ".join(tokens)])

print("Vocabulary:")
print(cv.get_feature_names_out())

print("BoW Matrix:")
print(bow.toarray())

# =====================================================
# TF-IDF
# =====================================================
print("\n========== TF-IDF ==========")

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform([" ".join(tokens)])

print("Vocabulary:")
print(tfidf.get_feature_names_out())

print("TF-IDF Matrix:")
print(tfidf_matrix.toarray())

# =====================================================
# POS Tagging
# =====================================================
print("\n========== POS TAGGING ==========")

doc = nlp(" ".join(tokens))

for token in doc:
    print(token.text, "->", token.pos_)

# =====================================================
# Named Entity Recognition (NER)
# =====================================================
print("\n========== NAMED ENTITY RECOGNITION ==========")

if doc.ents:
    for ent in doc.ents:
        print(ent.text, "->", ent.label_)
else:
    print("No Named Entities Found.")
```
