```python
import numpy as np
import pandas as pd
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras import Sequential
```

```python
train = pd.read_csv("DailyDelhiClimateTrain.csv")
test = pd.read_csv("DailyDelhiClimateTrain.csv")
```

```python
train = train["meantemp"].values
test = test["meantemp"].values
```

```python
x_train = []
y_train = []

for i in range(len(train)-5):
    x_train.append(train [i:i+5])
    y_train.append(train[i+5])
```

```python
x_test = []
y_test = []
for i in range(len(train)-5):
    x_test.append(test [i:i+5])
    y_test.append(test [i+5])
```

```python
x_train = np.array(x_train)
y_train = np.array(y_train)
x_test = np.array(x_test)
y_test = np.array(y_test)

x_train = x_train.reshape(x_train.shape[0],5,1)
x_test = x_test.reshape(x_test.shape[0],5,1)
```

```python
model = Sequential ([
    LSTM(32,input_shape = (5,1)),
    Dense(1)
])
```

```python
model.compile(
    optimizer = "adam",
    loss = "mse"
)
```

```python
model.fit( x_train, y_train ,epochs = 50,batch_size = 32)
```

```python
predicted = model.predict(x_test[:1])
```

```python
print("result")
print("actual:",y_test[0])
print("predicted", predicted [0][0])
```

```python
data = pd.read_csv("monthly-milk-production-pounds.csv")
```

```python
milk = data["Price"].dropna().values
```

```python
maximum = max(milk)
print(maximum)
```

```python
milk = milk / maximum
```

```python
x = []
y = []

for i in range(len(milk) - 5):
    x.append(milk[i:i+5])
    y.append(milk[i+5])
```

```python
x = np.array(x)
y = np.array(y)
```

```python
x = x.reshape(x.shape[0],5,1)
```

```python
model = Sequential([

    LSTM(32,input_shape=(5,1)),
         Dense(1)
])
```

```python
model.compile(
    optimizer = "adam",
    loss = "mse"
)
```

```python
model.fit(x,y,epochs=20,verbose = 0,batch_size=8)
```

```python
predicted = model.predict(x[:1])
```

```python
print("actual:",y[0] * maximum)
print("predicted",predicted[0][0] * maximum)
```
