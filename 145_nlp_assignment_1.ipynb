```python
import spacy

# Task 1: Load English model
nlp = spacy.load("en_core_web_sm")

# Dataset
text = """
OpenAI developed ChatGPT in San Francisco.
Dr. ABC teaches at GLS University, Ahmedabad.
Google launched Gemini in 2024.
"""

# Task 2: Tokenize text
doc = nlp(text)

# Task 3 & 4: POS Tagging
print("POS TAGGING")
print("-" * 40)

noun_count = 0
verb_count = 0

for token in doc:
    # Ignore spaces and punctuation
    if not token.is_space and not token.is_punct:
        print(f"Token: {token.text:15} POS: {token.pos_}")

        # Count nouns (common + proper)
        if token.pos_ in ["NOUN", "PROPN"]:
            noun_count += 1

        # Count verbs
        if token.pos_ == "VERB":
            verb_count += 1

# Task 5 & 6: Named Entity Recognition
print("\nNAMED ENTITY RECOGNITION")
print("-" * 40)

entity_count = len(doc.ents)

for ent in doc.ents:
    print(f"Entity: {ent.text:25} Label: {ent.label_}")

# Task 7: Counts
print("\nCOUNTS")
print("-" * 40)
print(f"Total Nouns: {noun_count}")
print(f"Total Verbs: {verb_count}")
print(f"Total Named Entities: {entity_count}")

# Task 8: Identify PERSON, ORG, GPE, DATE
print("\nSPECIFIC ENTITY TYPES")
print("-" * 40)

persons = []
orgs = []
gpes = []
dates = []

for ent in doc.ents:
    if ent.label_ == "PERSON":
        persons.append(ent.text)
    elif ent.label_ == "ORG":
        orgs.append(ent.text)
    elif ent.label_ == "GPE":
        gpes.append(ent.text)
    elif ent.label_ == "DATE":
        dates.append(ent.text)

print("PERSON:", persons)
print("ORG:", orgs)
print("GPE:", gpes)
print("DATE:", dates)
```
